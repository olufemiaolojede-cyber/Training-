```python
def find_max_min(numbers):
    maximum = numbers[0]
    minimum = numbers[0]

    for num in numbers:
        if num > maximum:
            maximum = num
        if num < minimum:
            minimum = num

    return maximum, minimum
```

```python
values = [45, 67, 89, 34, 90, 55, 72]
max_value, min_value = find_max_min(values)

print("Maximum:", max_value)
print("Minimum:", min_value)
```
